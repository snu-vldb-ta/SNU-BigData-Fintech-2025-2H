```python
def retrieve_documents(query, collection):
    results = collection.query(query_texts=[query], n_results=1)
    most_similar_doc = results['documents']

    return most_similar_doc
```

```python
def generate_response_RAG(query, relevant_document):
    prompt = f"Given the following document: '{relevant_document}', answer the query: '{query}'"

    response = openai_client.chat.completions.create(
    messages=[
        {
              "role": "user",
              "content": prompt,
          }
      ],
      model="gpt-4o",
    )

    return response.choices[0].message.content
```

```python
def generate_response_llm(query):
    prompt = f"Answer the query: '{query}'"

    response = openai_client.chat.completions.create(
    messages=[
        {
              "role": "user",
              "content": prompt,
          }
      ],
      model="gpt-4o",
    )

    return response.choices[0].message.content
```

```python
def rag_system(query, collection):
    relevant_document = retrieve_documents(query, collection)
    response = generate_response_RAG(query, relevant_document)
    return response
```

```python
query = "Why is vitamin C important for health, and which fruits are high in it?"
response1 = generate_response_llm(query)
response2 = rag_system(query, collection)

print("[Query] \n", query)
print("=======================================================================")
print("[Response using LLM only] \n",response1)
print("=======================================================================")
print("[Response using RAG] \n",response2)

# other possible queries
# "What are the health benefits of oranges?"
# "How does eating grapes support cardiovascular health?"
# "What vitamins are found in kiwis and how do they benefit skin and immunity?"
# "Describe the taste and texture differences between apples and pears."
# "How is a banana useful for post-exercise recovery and energy replenishment?"
# "What makes a pomegranate unique among fruits in terms of nutrients?"
# "How are strawberries commonly used in modern and traditional recipes?"
# "In what ways are mangosteens and lychees similar in health benefits?"
# "What enzyme does pineapple contain and how does it help digestion?"
# "What are the advantages of healthy fats in avocados?"
# "Compare the immune-boosting properties of grapefruit and kiwi."
# "How do blueberries support brain and heart health?"
# "What is the culinary and nutritional use of figs across cultures?"
# "Which fruits are best for skin health and why?"
# "How does dietary fiber in apples, figs, and mulberries aid digestion?"
# "What tropical fruits are commonly used in Southeast Asian cuisine?"
# "Why is lychee considered both refreshing and energizing?"
# "What role does potassium play in bananas for muscle recovery?"
# "How do antioxidant levels compare across berries like blueberry and mulberry?"
# "What are the hydration benefits of fruits like pear and watermelon?"
```
